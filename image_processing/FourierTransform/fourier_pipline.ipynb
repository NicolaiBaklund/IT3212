```python
# Image Addition using add_images function
from task2_and_task3 import add_images
import cv2
import numpy as np
import os

# Define image paths
image_path = base_dir + "BilderFourier/41.jpg"
image_path2 = base_dir + "FourierGen/noisy_dots.jpg"

# Load the images
img1 = cv2.imread(image_path, cv2.IMREAD_GRAYSCALE)
img2 = cv2.imread(image_path2, cv2.IMREAD_GRAYSCALE)

print(f"Image 1 shape: {img1.shape}")
print(f"Image 2 shape: {img2.shape}")

# Resize images to the same size (use the smaller dimensions to avoid losing information)
min_height = min(img1.shape[0], img2.shape[0])
min_width = min(img1.shape[1], img2.shape[1])

print(f"Resizing both images to: {min_height}x{min_width}")

# Resize both images to the same dimensions
img1_resized = cv2.resize(img1, (min_width, min_height))
img2_resized = cv2.resize(img2, (min_width, min_height))

# Scale the intensity of pixel values in img2 (not img1)
# Scale factor for img2 intensity
intensity_scale = 0.7  # Adjust this value to control the intensity of img2

# Scale the pixel values of img2
img2_scaled = (img2_resized * intensity_scale).astype(np.uint8)

# Add the images (img1 unchanged, img2 with scaled intensity)
result = add_images(img1_resized, img2_scaled, scale_factor=1.0)

# Get the name of img1 for output filename
img1_name = os.path.basename(image_path).split('.')[0]  # Get filename without extension
out_path = base_dir + "BilderFourier/noisy_" + img1_name + ".jpg"

# Save the result
cv2.imwrite(out_path, result)

print(f"Result saved to: {out_path}")
print(f"Result image shape: {result.shape}")
print(f"Result pixel range: [{result.min()}, {result.max()}]")
```
